````python
# Start with imports - ask ChatGPT to explain any package that you don't know
# This lab uses Gemini via the OpenAI-compatible API (see Guide 9)

import os
import re
import json
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import Markdown, display
````

````python
# Always remember to do this!
load_dotenv(override=True)
````

````python
# Check API keys - this lab requires GEMINI_API_KEY in your .env file (project root)

gemini_api_key = os.getenv('GEMINI_API_KEY')
deepseek_api_key = os.getenv('DEEPSEEK_API_KEY')
groq_api_key = os.getenv('GROQ_API_KEY')

if gemini_api_key:
    print(f"Gemini API Key exists and begins {gemini_api_key[:8]}")
else:
    print("Gemini API Key not set - add GEMINI_API_KEY to your .env file in the agents folder")

if deepseek_api_key:
    print(f"DeepSeek API Key exists and begins {deepseek_api_key[:3]}")
else:
    print("DeepSeek API Key not set (optional - extra competitor skipped)")

if groq_api_key:
    print(f"Groq API Key exists and begins {groq_api_key[:4]}")
else:
    print("Groq API Key not set (optional - extra competitor skipped)")
````

````python
request = "Please come up with a challenging, nuanced question that I can ask a number of LLMs to evaluate their intelligence. "
request += "Answer only with the question, no explanation."
messages = [{"role": "user", "content": request}]
````

````python
messages
````

````python
# Gemini via OpenAI-compatible endpoint (same pattern as lab 1)
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=gemini_api_key)

response = gemini.chat.completions.create(
    model="gemini-2.5-flash-lite",
    messages=messages,
)
question = response.choices[0].message.content
print(question)
````

````python
competitors = []
answers = []
messages = [{"role": "user", "content": question}]
````

## Note - Gemini edition

This notebook uses **GEMINI_API_KEY** from your `.env` file. The main competitors are different Gemini models (fast and capable). Optional cells below add DeepSeek, Groq, or Ollama if you have those set up.

````python
# Competitor 1 - fast Gemini model

model_name = "gemini-2.5-flash-lite"

response = gemini.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
````

````python
# Competitor 2 - more capable Gemini model

model_name = "gemini-2.5-flash"

response = gemini.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
````

````python
# Competitor 3 - another Gemini variant for comparison

model_name = "gemini-2.0-flash"

response = gemini.chat.completions.create(model=model_name, messages=messages)
answer = response.choices[0].message.content

display(Markdown(answer))
competitors.append(model_name)
answers.append(answer)
````

````python
# Optional - only runs if DEEPSEEK_API_KEY is in your .env

if deepseek_api_key:
    deepseek = OpenAI(api_key=deepseek_api_key, base_url="https://api.deepseek.com/v1")
    model_name = "deepseek-chat"

    response = deepseek.chat.completions.create(model=model_name, messages=messages)
    answer = response.choices[0].message.content

    display(Markdown(answer))
    competitors.append(model_name)
    answers.append(answer)
else:
    print("Skipping DeepSeek - no API key")
````

````python
# Optional - only runs if GROQ_API_KEY is in your .env

if groq_api_key:
    groq = OpenAI(api_key=groq_api_key, base_url="https://api.groq.com/openai/v1")
    model_name = "openai/gpt-oss-120b"

    response = groq.chat.completions.create(model=model_name, messages=messages)
    answer = response.choices[0].message.content

    display(Markdown(answer))
    competitors.append(model_name)
    answers.append(answer)
else:
    print("Skipping Groq - no API key")
````

## For the next cell, we will use Ollama

and runs models locally using high performance C++ code.

You might need to restart Cursor (and maybe reboot). Then open a Terminal (control+\`) and run `ollama serve`

Useful Ollama commands (run these in the terminal, or with an exclamation mark in this notebook):

`ollama pull <model_name>` downloads a model locally  
`ollama ls` lists all the models you've downloaded  
`ollama rm <model_name>` deletes the specified model from your downloads

````python
# So where are we?

print(competitors)
print(answers)
````

````python
# It's nice to know how to use "zip"
for competitor, answer in zip(competitors, answers):
    print(f"Competitor: {competitor}\n\n{answer}")
````

````python
# Let's bring this together - note the use of "enumerate"

together = ""
for index, answer in enumerate(answers):
    together += f"# Response from competitor {index+1}\n\n"
    together += answer + "\n\n"
````

````python
print(together)
````

````python
judge = f"""You are judging a competition between {len(competitors)} competitors.
Each model has been given this question:

{question}

Your job is to evaluate each response for clarity and strength of argument, and rank them in order of best to worst.

Each response is labeled "# Response from competitor N" where N is 1, 2, 3, etc.

Respond with JSON only (no markdown), using integers for competitor numbers:
{{"results": [2, 1, 3]}}

The list must contain each competitor number from 1 to {len(competitors)} exactly once, ordered best to worst.
Use integers only — not strings like "competitor 2".

Here are the responses from each competitor:

{together}

Now respond with the JSON ranked order, nothing else."""
````

````python
print(judge)
````

````python
judge_messages = [{"role": "user", "content": judge}]
````

````python
# Judgement time! Gemini judges the competition

response = gemini.chat.completions.create(
    model="gemini-2.5-flash",
    messages=judge_messages,
)
results = response.choices[0].message.content
print(results)
````

````python
# OK let's turn this into results!

def competitor_number(value):
    """Parse 1-based competitor number from int or strings like 'competitor 2'."""
    if isinstance(value, int):
        return value
    match = re.search(r"\d+", str(value))
    if not match:
        raise ValueError(f"Could not parse competitor number from: {value!r}")
    return int(match.group())

results_text = results.strip()
if results_text.startswith("```"):
    results_text = re.sub(r"^```(?:json)?\s*|\s*```$", "", results_text, flags=re.MULTILINE).strip()

results_dict = json.loads(results_text)
ranks = results_dict["results"]
for index, result in enumerate(ranks):
    competitor = competitors[competitor_number(result) - 1]
    print(f"Rank {index+1}: {competitor}")
````
